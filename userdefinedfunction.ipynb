```python
#function Without parameters
def greet():
    print("hello!!!!! WELCOME to python.")
greet()
```

```python
#function with parameter
def user(name):
    print("hello",name)
user("Akash")
```

```python
#function with two parameters
def add(a,b):
    print("sum=",a+b)
add(10,20)
```

```python
#function with return value
def multiply(a,b):
    return a*b
result=multiply(5,4)
print("multiplication=",result)
```

```python
#function with default parameter
def country(name="india"):
    print("country",name)
country()
country("japan")
print()
```

```python
#function to check even or odd
def check_even_odd(num):
    if num%2==0:
        print("num is even=",num)
    else:
        print("num is odd=",num)
check_even_odd(5)
check_even_odd(10)
```

```python
#function to find maximum
def maximum(a,b):
    if a>b:
        return a
    else:
        return b
print("maximum=",maximum(2,18))
```

```python
#function to calculate square
def square(num):
    return num**2
print("square=",square(20))
print()
```

```python
#function to calculate factorial
def factorial(n):
    fact=1
    for i in range(1,n+1):
        fact *=i
    return fact
print("factorial of 5 is",factorial(5))
print()
```

```python
#function to check positive,negative,zero
def check_num(num):
    if num>0:
        print("positive")
    elif num<0:
        print("negative")
    else:
        print("zero")
check_num(22)
print()
```

```python
#modulus/reminder
def mod(n1,n2):
    res=n1%n2
    return res
total=mod(100,3)
print(total)
```
