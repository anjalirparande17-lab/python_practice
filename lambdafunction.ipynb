# lambda functions / anonymous function

```python
#===========syntax-lambda arguments:expression===================
```

```python
#addtion
res=lambda a,b:a+b
res(4,3)
```

```python
#square
square=lambda x:x**2
square(23)
```

```python
# find the largest number
maximum=lambda a,b:a if a>b else b
print(maximum(20,22))
```

```python
# check even or odd
even_odd=lambda x: "even" if x%2==0 else "odd"
print(even_odd(8))
print(even_odd(7))
```

```python
# check positive negative
check=lambda x: "positive" if x>0 else "negative"
print(check(10))
print(check(-1100))
```

```python
# minimum
minimum=lambda a,b:a if a <b else b
print(minimum(10,11))
```

```python
#square root
sqrt=lambda x:x**(1/2)
sqrt(25)
```

```python
#area of circle
CA=lambda r:3.14*r**2
CA(2.5)
```

```python
#area of triangle
Aot=lambda h,b:h*b
Aot(20,3)
```

```python
#filter -it is a function used for filtering the value in collection.
#syntax
#filter(function,collection)
```

```python
l1=[2,3,4,5,6,7,8,90,1]
```

```python
#even number print
def EV(x):
    if x%2==0:
        return True
    else:
        return False
```

```python
filter(EV,l1)
```

```python
list(filter(EV,l1))
```

```python
#map
#syntax
#map(function,collection)
```

```python
l1=[2,3,4,5,6,7,8,90,1]
```

```python
list(map(lambda x:x**0.5,l1))
```
