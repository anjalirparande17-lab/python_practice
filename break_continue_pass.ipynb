# Break Statement


```python
for i in range (1,11):
    if i==6:
        break
    print(i)
```

```python
for ch  in 'python':
    if ch=="h":
        break
    print(ch)
```

# continue statement


```python
for i in range(1,11):
     if i==6:
         continue
     print(i)
```

```python
for i in range (1,11):
    if i%2==0:
        continue
    print(i)
```

# pass statement


```python
for i in range (1,6):
    if i==3:
        pass 
    print(i)
```

```python
# stop the loop when number is 8.
for i in range(1,11):
    if i==8:
        break
    print(i)
```

```python
#print only odd numbers
for i in range(1,21):
    if i%2==0:
        continue
    print(i)
```

```python
#skip multiples of 3
for i in range(1,21):
    if i%3==0:
        continue
    print(i)
```

```python
#find first even  number
numbers=[1,3,5,8,9,10]
for num in numbers:
    if num%2==0:
        print("first even number:",num)
        break
```

```python
#print vowels only
word="programming"
for ch in word:
    if ch not in "aeiou":
        continue
    print(ch)
```

```python
#break the  nested loop
for i in range (1,4):
    for j in range (1,6):
        if j==4:
            break
        print(i,j)
```

```python
#empty function using pass
def login():
    pass
print("login function created")
```

```python
#skip negative numbers 
numbers =[10,-2,-3,40,5,6,78]
for num in numbers:
    if num<0:
        continue
    print(num)
```

```python
#print numbers except 5 and 8
for i in range(1,11):
    if i==5 or i==8:
        continue
    print(i)
```
