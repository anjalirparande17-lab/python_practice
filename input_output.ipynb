# input output in python

```python
#example1-print output
print("hello world!")
print("welcome to python programming")
```

```python
#example2-taking string input
name=input("enter your name:")
print("hello",name)
```

```python
#example3-taking integer input
age=int(input("enter your age:"))
print("your age is:",age)
```

```python
#example4-taking float number
height=float(input("enter your height(in meters:)"))
print("your height is:",height)
```

```python
#example5-taking multiple inputs
a,b=input("Enter two numbers:").split()
print("first number:",a)
print("second number:",b)
```

```python
#example6-integer multiple inputs
x,y=map(int,input("Enter two integers:").split())
print("Sum=",x+y)
```

```python
#example7-formatted output
city=input("Enter your city")
print(f"you live in {city}.")
```

```python
#example8-printing multiple values
name="anjali"
course="MCA"
print(name,course)
```

```python
#example9-simple calculation
print("python","data","science",sep="-")
print("hello",end=" ")
print("world")
```

```python
#example10- simple calculation
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
```
