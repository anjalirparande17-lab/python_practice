# Type conversion in Python


```python
# integer to float 
num=10
print(float(num))
```

```python
# float to interger
num=10.34
print(int(num))
```

```python
# interger to string
age=21
print(str(age))
```

```python
#string to integer 
value="100"
print(int(value))
```

```python
# string to float
value="33.55"
print(float(value))
```

```python
#integer to boolean
number=1
print(bool(number))
```

```python
#integer to boolean
number=0
print(bool(number))
```

```python
#user input conversion
marks=input("Enter your marks:")
marks=float(marks)
```

```python
print("marks:",marks)
print("Data type:",type(marks))
```

```python
#integer to float
num=10
result=float(num)
print("\n1. Integer to float")
print("Original value:",num)
print("Datatype:",type(result))
```

```python
#float to integer
num=33.65
result=int(num)
print("\n1. float to integer")
print("Original value:",num)
print("Datatype:",type(result))
```

```python
# string to integer
number="100"
result=int(number)
print("string to integer")
print("Original value:",number)
print("Datatype:",type(result))
```

```python
#string to float
number="2.3"
result=float(number)
print("String to float")
print("Original value:",number)
print("Datatype",type(result))
```

```python
#float to string
salary=25000.00
result=str(salary)
print("float to string")
print("original value:",salary)
print("Converted value:",result)
print("Datatype:",type(result))
```

```python
#boolean to integer
x=True
y=False
print(" Boolean to Integer")
print("True=",int(x))
print("False=",int(y))
```
