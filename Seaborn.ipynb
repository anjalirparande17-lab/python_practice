```python
import warnings
warnings.filterwarnings("ignore")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
sns.get_dataset_names()
```

```python
# loading  dataset
df= sns.load_dataset("iris")
```

```python
df
```

```python
df.species.value_counts()
```

```python
df.species.unique()
```

```python
df.head(1)
```

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df,ci=68)
plt.show()
```

```python
sns.lineplot(x='sepal_length',y='sepal_width',data=df,hue='species',ci=None)
plt.show()
```

```python
sns.scatterplot(x='sepal_length',y='sepal_width',data=df,hue='species')
plt.title("Scatter plot")
plt.show()
```

```python
sns.barplot(x='species',y='sepal_length',data=df)
plt.show()
```

```python
sns.countplot(x='species', data=df)
plt.show()
```

```python
sns.boxplot(x='sepal_width',data=df)
plt.show()
```

```python
sns.boxplot(x='species',y='sepal_width',data=df)
plt.show()
```

```python
sns.violinplot(x="species",y='sepal_width',data=df)
plt.show()
```

```python
sns.stripplot(x="species",y='petal_width',data=df)
plt.show()
```

```python
sns.histplot(x="petal_width",data=df)
plt.show()
```

```python
sns.histplot(x="petal_width",data=df,hue='species',kde=True)
plt.show()
```

```python
sns.distplot(df['petal_width'],color="g")
plt.show()
```

```python
sns.pairplot(data=df,hue='species')
plt.show()
```

```python
# categorical plot
sns.catplot(x='sepal_length',data=df,kind='violin')
plt.show()
```

```python
sns.catplot(x='species', y='sepal_length', data=df, kind='swarm')
plt.show()
```

```python
sns.catplot(x='petal_length',y='species',data=df)
plt.show()
```

```python
tc=df.select_dtypes(include=['number']).corr()
print(tc)
```

```python
plt.figure(figsize=(7,8))
sns.heatmap(tc,annot=True,cmap='coolwarm')
plt.title('correlation heatmap')
plt.show()
```

```python
sns.heatmap(tc,annot=True)
plt.show()
```
